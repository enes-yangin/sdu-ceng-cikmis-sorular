# **Note**

Due to the approximate nature of numerical calculations, **accumulated errors** may lead to **extra iterations**. This is expected behavior, as each iteration refines the result but might require additional steps to achieve the desired accuracy. The convergence might thus take slightly longer than anticipated, especially in cases with small `eps` values or complex functions.


```python
def derivative_1(x, f, h=1e-5):
    return ( f( x + h ) - f( x - h) ) / (2 * h)

def derivative_2(x0, x1, f):
    return (f(x1) - f(x0)) / (x1 - x0)

def first_derivative(x0, f, h=1e-5):
    return ( f( x0 + h ) - f(x0) ) / h

def first_derivative_2(x0, x1, f, h=1e-6):
    f_prime_x0 = first_derivative(x0, f, h)
    f_prime_x1 = first_derivative(x1, f, h)   
    return (f_prime_x1 - f_prime_x0) / ( x1 - x0 )

def second_derivative(x0, f, h=1e-6, h1=1e-7):
    f_prime_x0 = first_derivative(x0, f, h)
    f_prime_x1 = first_derivative(x0 + h1, f, h)   
    return (f_prime_x1 - f_prime_x0) / h1
```

```python
def newton_method(x0, f, eps=1e-3, max_tries=100):
    iter = 0
    while iter < max_tries:
        print(f"x{iter} = {x0}")
        f_x0_p = first_derivative(x0, f)
        f_x0_pp = second_derivative(x0, f)
        
        xn = x0 - (f_x0_p / f_x0_pp)
        if abs(f_x0_p) <= eps:
            print(f"Extremum found after {iter + 1} iterations")
            print(f"x{iter + 1} = {xn}")
            return xn
        x0 = xn
        iter += 1
    print("Maximum iterations reached without convergence.")
    return None
```

```python
def gradient_descent(x0, f, alpha=1e-3, eps=1e-3, max_tries=100):
    iter = 0
    while iter < max_tries:
        
        if iter < 10:
            print(f"x{iter} = {x0}")
        
        f_x0_prime = first_derivative(x0, f)
        
        if abs(f_x0_prime) <= eps:
            print("...")
            print(f"Local minimum found after {iter} iterations")
            print(f"x{iter} = {x0}")
            return x0
        
        xn = x0 - alpha * f_x0_prime

        x0 = xn
        iter += 1
    print("Maximum iterations reached without convergence.")
    return None    
```

```python
def bisection_method(a, b, f, eps=1e-3, max_tries=100):
    if f(a) * f(b) > 0:
        print("Wrong range: f(a) and f(b) must have opposite signs. Try again with correct ranges.")
        return None, a, b

    iter = 0
    while iter < max_tries:
        x = (a + b) / 2
        f_x = f(x)

        if iter < 5:
            print(f"x{iter} = {x}")
            print(f"f(x{iter}) = {f_x}")
            print(f"{iter}. range [{a}, {b}]")

        if f_x == 0:
            print("...")
            print(f"Root found exactly at x{iter} = {x}")
            return x, a, b

        f_a = f(a)
        if f_a * f_x < 0:
            b = x
        else:
            a = x

        if abs(b - a) / 2 <= eps:
            print("...")
            print(f"Root found after {iter + 1} iterations")
            print(f"x{iter + 1} = {x}")
            return x, a, b

        iter += 1

    print("Maximum iterations reached without convergence.")
    return None, a, b
```

```python
def secant_method(x0, x1, f, eps=1e-3, max_tries=100, h=1e-3):
    iter = 0
    while iter < max_tries:
        f_x0_p = first_derivative(x0, f, h=h)
        
        if abs(x0 - x1) < eps:
            return x1
        
        f_x01_pp = first_derivative_2(x0, x1, f, h=h)

        if iter < 5:
            print(f"x{iter} = {x0}\nx{iter+1} = {x1}\n")

        xn = x0 - ( f_x0_p / f_x01_pp )

        if abs(f_x0_p) < eps:
            print("...")
            print(f"Extremum found after {iter + 2} iterations")
            print(f"x{iter} = {x0}\nx{iter+1} = {x1}\n")
            print(f"x{iter + 2} = {xn}")
            return xn
        
        x0 = x1
        x1 = xn
        iter += 1
    
    print("Maximum iterations reached without convergence.")
    return None
```

```python
import math

def newton_method_with_backtracking(x0, a, b, f, eps=1e-5, max_tries=100):
    # Ensure the starting point x0 is within the valid range [a, b]
    if not a < x0 < b:
        print(f"Choose valid x0 in range [{a}, {b}]")
        return None

    k = 0
    xk = x0
    fxk = f(xk)
    delta_x = math.inf

    while k < max_tries:
        if abs(fxk) <= eps or abs(delta_x) <= eps:
            return xk
        else:
            delta_x = fxk / first_derivative(xk, f)
            t = xk - delta_x
            ft = f(t)
            alpha = 1
            while not abs(ft) < abs(fxk):
                alpha = alpha * 0.5
                t = xk - alpha * delta_x
                ft = f(t)
            delta_x = xk - t
            xk = t
            fxk = ft
        k += 1
```

```python
import math

def safeguarded_newton_method(x0, a, b, f, eps=1e-5, max_tries=100):
    # Ensure the starting point x0 is within the valid range [a, b]
    if not a <= x0 <= b:
        print(f"Choose valid x0 in range [{a}, {b}]")
        return None

    k = 0
    xk = x0
    fxk = f(xk)
    fp_xk = first_derivative(xk, f)
    fpp_xk = second_derivative(xk, f)
    delta_x = math.inf
    alpha = 1

    while k < max_tries:
        # Debug output for each iteration
        print("\n" + " " * 40 + f" Iteration {k} " + " " * 40)
        print(100 * "#")
        print(f"x = {round(xk, 3)}, f(x) = {round(fxk, 3)}, f'(x) = {round(fp_xk, 3)}, f''(x) = {round(fpp_xk, 3)}, delta_x = {round(delta_x, 3)}, alpha = {round(alpha, 3)}")

        # Check for convergence to a root
        if abs(fp_xk) <= eps:
            print("Convergence reached at a root.")
            return xk

        # Calculate the Newton step
        if fpp_xk > 0:  # Ensure we only apply the Newton step when the second derivative is non-zero
            delta_x = -fp_xk / fpp_xk
        else:  # If the second derivative is zero, use a steepest descent step
            delta_x = -fp_xk

        t = xk + alpha * delta_x
        ft = f(t)
        fp_t = first_derivative(t, f)
        fpp_t = second_derivative(t, f)
        print(f"t = {round(t, 3)}, f(t) = {round(ft, 3)}, f'(x) = {round(fp_t, 3)}, f''(x) = {round(fpp_t, 3)}, delta_x = {round(delta_x, 3)}, alpha = {round(alpha, 3)}")

        # Ensure alpha stays within the interval [a, b]
        if delta_x < 0:
            alpha = min(1, (a - xk) / delta_x)
        elif delta_x > 0:
            alpha = min(1, (b - xk) / delta_x)

        print("################")
        # Reduce alpha if needed until the step decreases |f(x)|
        j = 0
        while True:
            print(f"J Iteration: {j}")
            print(f"Alpha = {alpha}")
            t = xk + alpha * delta_x
            fxk = f(xk)
            ft = f(t)
            if ft < fxk:
                print(f"f({t}) < f({xk})")
                print(f"{ft} < {fxk}")
                break
            else:
                print(f"f({t}) > f({xk})")
                print(f"{ft} > {fxk}")
            
            alpha *= 0.5
            j += 1
        
        print(f"Alpha = {alpha}")
        print("################")
        print(f"t = {round(t, 3)}, f(t) = {round(ft, 3)}, f'(x) = {round(fp_t, 3)}, f''(x) = {round(fpp_t, 3)}, delta_x = {round(delta_x, 3)}, alpha = {round(alpha, 3)}")
        # Update xk and check new values
        xk = t
        fxk = ft
        fp_xk = first_derivative(xk, f)
        fpp_xk = second_derivative(xk, f)
        print(100 * "#")
        
        k += 1

    print("Maximum number of iterations reached without convergence.")
    return None
```
