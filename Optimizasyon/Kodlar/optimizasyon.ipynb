```python
import numpy as np

def f(x):
    # Amaç fonksiyonumuz
    return x**3 - 3*x**2

def f_prime(x):
    # Fonksiyonun türevi
    return 3*x**2 - 6*x

def f_double_prime(x):
    # Fonksiyonun ikinci türevi
    return 6*x - 6

def newton_method(x0, tol=1e-5, max_iter=100):
    """Standart Newton Yöntemi"""
    x = x0
    for i in range(max_iter):
        fx = f(x)
        fpx = f_prime(x)
        fppx = f_double_prime(x)
        
        if abs(fpx) < tol:  # Gradyan küçükse dur
            print(f"Newton yöntemi durduruldu: iterasyon={i}, x={x}, f(x)={fx}")
            return x, fx
        
        delta_x = -fpx / fppx  # Newton adımı
        x += delta_x
        
        if abs(delta_x) < tol:
            print(f"Newton yöntemi başarılı: iterasyon={i}, x={x}, f(x)={fx}")
            return x, fx
        
    print("Newton yöntemi maksimum iterasyon sayısına ulaştı.")
    return x, f(x)

def safeguarded_newton_method(x0, tol=1e-5, max_iter=100):
    """Korumalı Newton Yöntemi"""
    x = x0
    for i in range(max_iter):
        fx = f(x)
        fpx = f_prime(x)
        fppx = f_double_prime(x)
        
        if abs(fpx) < tol:  # Gradyan küçükse dur
            print(f"Korumalı Newton durduruldu: iterasyon={i}, x={x}, f(x)={fx}")
            return x, fx
        
        delta_x = -fpx / fppx  # Newton adımı
        
        # Korumalı adım: eğer adım çok büyükse küçült
        alpha = 1.0
        while abs(f(x + alpha * delta_x)) > abs(fx):
            alpha *= 0.5  # Adımı yarıya indirerek ilerle
        
        x += alpha * delta_x
        
        if abs(alpha * delta_x) < tol:
            print(f"Korumalı Newton başarılı: iterasyon={i}, x={x}, f(x)={fx}")
            return x, fx
        
    print("Korumalı Newton yöntemi maksimum iterasyona ulaştı.")
    return x, f(x)

# Başlangıç değeri
x0 = 1.1

# Standart Newton yöntemi
print("Newton Yöntemi Sonuçları:")
newton_method(x0)

# Korumalı Newton yöntemi
print("\nKorumalı Newton Yöntemi Sonuçları:")
safeguarded_newton_method(x0)
```

# Assistant
Verdiğiniz kod, bir matematiksel fonksiyonun köklerini bulmak için Newton Yöntemi ve Korumalı Newton Yöntemi'ni kullanmaktadır. Bu yöntemler, bir fonksiyonun sıfır olduğu noktaları (yani köklerini) bulmak için iteratif bir yaklaşım kullanır.

### Denklemin Açıklaması

Kodda tanımlanan fonksiyon şu şekildedir:

\[ f(x) = x^3 - 3x^2 \]

Bu denklem, bir polinom fonksiyonudur ve x'in değerine bağlı olarak farklı sonuçlar üretir. Fonksiyonun köklerini bulmak için, bu denklemin sıfıra eşit olduğu noktaları arıyoruz:

\[ x^3 - 3x^2 = 0 \]

### Türevler

Fonksiyonun türevleri de tanımlanmıştır:

1. **Birinci Türev**:
   \[ f'(x) = 3x^2 - 6x \]
   Bu, fonksiyonun eğimini temsil eder ve kök bulma sürecinde kullanılır.

2. **İkinci Türev**:
   \[ f''(x) = 6x - 6 \]
   Bu, fonksiyonun eğiminin değişim hızını gösterir ve Korumalı Newton Yöntemi'nde adım büyüklüğünü ayarlamak için kullanılır.

### Yöntemler

1. **Newton Yöntemi**: 
   - Başlangıç değeri (x0) ile başlar ve her iterasyonda, bir adım hesaplayarak köke yaklaşır.
   - Eğer türev (f') sıfıra çok yakınsa, iterasyonu durdurur.

2. **Korumalı Newton Yöntemi**:
   - Aynı şekilde başlar, ancak adım büyüklüğünü korumak için bir kontrol mekanizması ekler.
   - Eğer adım çok büyükse, adımı yarıya indirerek daha güvenli bir yaklaşım sağlar.

### Sonuç

Bu kod, başlangıç değeri olarak \( x_0 = 1.1 \) kullanarak her iki yöntemi de çalıştırır ve kök bulma sürecinin sonuçlarını ekrana yazdırır. 

Eğer daha fazla bilgi veya belirli bir konu hakkında detay isterseniz, lütfen belirtin!

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

def newton_method(f, f_prime, f_double_prime, x0, max_iter=10, tol=1e-5):
    """Standart Newton Yöntemi"""
    x = x0
    results = []
    for i in range(max_iter):
        fx = f(x)
        fpx = f_prime(x)
        fppx = f_double_prime(x)
        
        if abs(fpx) < tol:
            results.append((i, x, fx, "DURDUR"))
            break
        
        delta_x = -fpx / fppx
        x += delta_x
        results.append((i, x, fx, delta_x))
        
        if abs(delta_x) < tol:
            results.append((i, x, fx, "BAŞARI"))
            break
    return results

def safeguarded_newton_method(f, f_prime, f_double_prime, x0, max_iter=10, tol=1e-5):
    """Korumalı Newton Yöntemi"""
    x = x0
    results = []
    for i in range(max_iter):
        fx = f(x)
        fpx = f_prime(x)
        fppx = f_double_prime(x)
        
        if abs(fpx) < tol:
            results.append((i, x, fx, "DURDUR"))
            break
        
        delta_x = -fpx / fppx
        alpha = 1.0
        while abs(f(x + alpha * delta_x)) > abs(fx):
            alpha *= 0.5
        x += alpha * delta_x
        results.append((i, x, fx, alpha * delta_x))
        
        if abs(alpha * delta_x) < tol:
            results.append((i, x, fx, "BAŞARI"))
            break
    return results

def main():
    # Kullanıcıdan fonksiyon ve parametreleri al
    x = sp.symbols('x')
    function_expr = input("Denklemi girin (örneğin x**3 - 3*x**2): ")
    f_sympy = sp.sympify(function_expr)
    
    f_prime_sympy = sp.diff(f_sympy, x)
    f_double_prime_sympy = sp.diff(f_prime_sympy, x)

    f = sp.lambdify(x, f_sympy, 'numpy')
    f_prime = sp.lambdify(x, f_prime_sympy, 'numpy')
    f_double_prime = sp.lambdify(x, f_double_prime_sympy, 'numpy')
    
    x0 = float(input("Başlangıç değerini girin: "))
    max_iter = int(input("İterasyon sayısını girin: "))
    
    newton_results = newton_method(f, f_prime, f_double_prime, x0, max_iter)
    safeguarded_results = safeguarded_newton_method(f, f_prime, f_double_prime, x0, max_iter)
    
    # Iterasyon sonuçlarını yan yana ekrana bas
    print(f"\n{'Iterasyon':<10} {'Newton x':<15} {'Newton f(x)':<15} {'Adım':<10} {'Korumalı Newton x':<20} {'Korumalı Newton f(x)':<20} {'Adım':<10}")
    print("="*95)
    
    for i in range(max(len(newton_results), len(safeguarded_results))):
        newton_data = newton_results[i] if i < len(newton_results) else ("-", "-", "-", "-")
        safeguarded_data = safeguarded_results[i] if i < len(safeguarded_results) else ("-", "-", "-", "-")
        
        print(f"{newton_data[0]:<10} {newton_data[1]:<15.5f} {newton_data[2]:<15.5f} {newton_data[3]:<10} "
              f"{safeguarded_data[1]:<20.5f} {safeguarded_data[2]:<20.5f} {safeguarded_data[3]}")
    
    # Grafik çizimi
    newton_x_values = [x[1] for x in newton_results]
    safeguarded_x_values = [x[1] for x in safeguarded_results]
    iterations = list(range(len(newton_x_values)))
    
    plt.plot(iterations, newton_x_values, 'o-', label="Newton Yöntemi", color='blue')
    plt.plot(iterations, safeguarded_x_values, 's-', label="Korumalı Newton Yöntemi", color='orange')
    plt.xlabel("İterasyon")
    plt.ylabel("x Değeri")
    plt.legend()
    plt.title("Newton ve Korumalı Newton Yöntemleri Karşılaştırması")
    plt.grid()
    plt.show()

main()
```

# secant yöntemi

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

def secant_method(f_prime, x0, x1, tol=1e-5):
    """Secant Yöntemi - yeterince küçük türev değeri sağlanana kadar iterasyon yapar"""
    results = [(0, x0, f_prime(x0))]
    results.append((1, x1, f_prime(x1)))
    
    iter_count = 2
    while True:
        fpx0 = f_prime(x0)
        fpx1 = f_prime(x1)
        
        if abs(fpx1 - fpx0) < tol:
            results.append((iter_count, x1, fpx1, "DURDUR"))
            break
        
        x2 = x0 - fpx0 * (x1 - x0) / (fpx1 - fpx0)
        delta_x = x2 - x1
        results.append((iter_count, x2, f_prime(x2), delta_x))
        
        if abs(f_prime(x2)) < tol:
            results.append((iter_count, x2, f_prime(x2), "BAŞARI"))
            break
        
        x0, x1 = x1, x2  # x_{k+2} için önceki iki değeri güncelle
        iter_count += 1
    
    return results

def main():
    # Kullanıcıdan fonksiyon ve parametreleri al
    x = sp.symbols('x')
    function_expr = input("Denklemi girin (örneğin x**3 - 3*x**2): ")
    f_sympy = sp.sympify(function_expr)
    f_prime_sympy = sp.diff(f_sympy, x)

    f_prime = sp.lambdify(x, f_prime_sympy, 'numpy')
    
    x0 = float(input("Başlangıç değeri x0 girin: "))
    x1 = float(input("Başlangıç değeri x1 girin: "))
    tol = float(input("Tolerans değerini girin (örneğin 1e-5): "))
    
    secant_results = secant_method(f_prime, x0, x1, tol)
    
    # Iterasyon sonuçlarını ekrana bas
    print(f"\n{'Iterasyon':<10} {'Secant x':<15} {'Secant f\'(x)':<15} {'Adım':<10}")
    print("="*50)
    
    for data in secant_results:
        if len(data) == 4:
            print(f"{data[0]:<10} {data[1]:<15.5f} {data[2]:<15.5f} {data[3]}")
        else:
            print(f"{data[0]:<10} {data[1]:<15.5f} {data[2]:<15.5f} {'-'}")
    
    # Grafik çizimi
    x_values = [x[1] for x in secant_results]
    iterations = list(range(len(x_values)))
    
    plt.plot(iterations, x_values, 'o-', label="Secant Yöntemi", color='purple')
    plt.xlabel("İterasyon")
    plt.ylabel("x Değeri")
    plt.legend()
    plt.title("Secant Yöntemi ile Kök Bulma")
    plt.grid()
    plt.show()

main()

```

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

def modified_secant_method(f_prime, x0, x1, tol=1e-5):
    """Değiştirilmiş Secant Yöntemi - yeterince küçük türev değeri sağlanana kadar iterasyon yapar
       Her adımda x0 sabit tutularak kullanılır."""
    initial_x0 = x0  # x0'ı sabit tutmak için başlangıç x0 değerini kaydediyoruz
    results = [(0, x0, f_prime(x0))]
    results.append((1, x1, f_prime(x1)))
    
    iter_count = 2
    while True:
        fpx0 = f_prime(initial_x0)  # x0 her iterasyonda sabit kalır
        fpx1 = f_prime(x1)
        
        if abs(fpx1 - fpx0) < tol:
            results.append((iter_count, x1, fpx1, "DURDUR"))
            break
        
        x2 = initial_x0 - fpx0 * (x1 - initial_x0) / (fpx1 - fpx0)
        delta_x = x2 - x1
        results.append((iter_count, x2, f_prime(x2), delta_x))
        
        if abs(f_prime(x2)) < tol:
            results.append((iter_count, x2, f_prime(x2), "BAŞARI"))
            break
        
        x1 = x2  # Sadece x1 güncellenir, x0 sabit kalır
        iter_count += 1
    
    return results

def main():
    # Kullanıcıdan fonksiyon ve parametreleri al
    x = sp.symbols('x')
    function_expr = input("Denklemi girin (örneğin x**3 - 3*x**2): ")
    f_sympy = sp.sympify(function_expr)
    f_prime_sympy = sp.diff(f_sympy, x)

    f_prime = sp.lambdify(x, f_prime_sympy, 'numpy')
    
    x0 = float(input("Başlangıç değeri x0 girin: "))
    x1 = float(input("Başlangıç değeri x1 girin: "))
    tol = float(input("Tolerans değerini girin (örneğin 1e-5): "))
    
    # Orijinal ve değiştirilen secant yöntemini çalıştır
    original_secant_results = secant_method(f_prime, x0, x1, tol)
    modified_secant_results = modified_secant_method(f_prime, x0, x1, tol)
    
    # Iterasyon sonuçlarını ekrana bas
    print(f"\n{'Iterasyon':<10} {'Orijinal Secant x':<20} {'Değiştirilmiş Secant x':<20}")
    print("="*60)
    
    max_iter = max(len(original_secant_results), len(modified_secant_results))
    for i in range(max_iter):
        orig_data = original_secant_results[i] if i < len(original_secant_results) else ("-", "-", "-")
        mod_data = modified_secant_results[i] if i < len(modified_secant_results) else ("-", "-", "-")
        
        orig_x = f"{orig_data[1]:<20.5f}" if isinstance(orig_data[1], (float, int)) else orig_data[1]
        mod_x = f"{mod_data[1]:<20.5f}" if isinstance(mod_data[1], (float, int)) else mod_data[1]
        
        print(f"{i:<10} {orig_x} {mod_x}")
    
    # Grafik çizimi
    orig_x_values = [x[1] for x in original_secant_results if isinstance(x[1], (float, int))]
    mod_x_values = [x[1] for x in modified_secant_results if isinstance(x[1], (float, int))]
    iterations = list(range(max(len(orig_x_values), len(mod_x_values))))
    
    plt.plot(iterations[:len(orig_x_values)], orig_x_values, 'o-', label="Orijinal Secant Yöntemi", color='blue')
    plt.plot(iterations[:len(mod_x_values)], mod_x_values, 's-', label="Değiştirilmiş Secant Yöntemi", color='red')
    plt.xlabel("İterasyon")
    plt.ylabel("x Değeri")
    plt.legend()
    plt.title("Orijinal ve Değiştirilmiş Secant Yöntemleri Karşılaştırması")
    plt.grid()
    plt.show()

def secant_method(f_prime, x0, x1, tol=1e-5):
    """Orijinal Secant Yöntemi - yeterince küçük türev değeri sağlanana kadar iterasyon yapar"""
    results = [(0, x0, f_prime(x0))]
    results.append((1, x1, f_prime(x1)))
    
    iter_count = 2
    while True:
        fpx0 = f_prime(x0)
        fpx1 = f_prime(x1)
        
        if abs(fpx1 - fpx0) < tol:
            results.append((iter_count, x1, fpx1, "DURDUR"))
            break
        
        x2 = x0 - fpx0 * (x1 - x0) / (fpx1 - fpx0)
        delta_x = x2 - x1
        results.append((iter_count, x2, f_prime(x2), delta_x))
        
        if abs(f_prime(x2)) < tol:
            results.append((iter_count, x2, f_prime(x2), "BAŞARI"))
            break
        
        x0, x1 = x1, x2  # x_{k+2} için önceki iki değeri güncelle
        iter_count += 1
    
    return results

main()

```
